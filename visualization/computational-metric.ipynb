```python
# 📈 All Calibration plots for Multiple Models on Specific Slide
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.calibration import calibration_curve
import numpy as np
import joblib
import pandas as pd
import warnings
warnings.filterwarnings('ignore')

def generate_calibration_plots_for_models(df, slide_num, n, nn):
    """
    Generate calibration plots for 10 different models on a specific slide
    
    Parameters:
    df: DataFrame with features and LABEL
    slide_num: Which slide to analyze (1-based indexing)
    n: Training data size parameter
    nn: Test data size parameter
    """
    print(f"Generating Calibration Curves for Slide {slide_num}...")
    
    # Model names and their corresponding file prefixes
    model_names = ['ET', 'RP', 'RotF', 'RRotF', 'AdaB', 'LB', 'GBM', 'XGB', 'CAT', 'LGB']
    
    probas_test_list = []
    probas_train_list = []
    y_tests = []
    y_trains = []
    clf_names = []
    
    # Calculate slide indices
    a = nn * (slide_num - 1)  # Convert to 0-based indexing for calculation
    
    for model_name in model_names:
        try:
            # Load model and selected features
            model = joblib.load(f"/kaggle/working/model_{model_name}_slide_{slide_num}.pkl")
            selected_features = joblib.load(f"/kaggle/working/selected_features_{model_name}_slide_{slide_num}.pkl")
            
            # Test data
            X_test_raw = df[selected_features][n+a+1:n+a+1+nn]
            y_test = df['LABEL'][n+a+1:n+a+1+nn]
            probas_test = model.predict_proba(pd.DataFrame(np.array(X_test_raw), columns=selected_features, index=X_test_raw.index))[:, 1]
            
            # Train data
            X_train_raw = df[selected_features][:n+a+1]
            y_train = df['LABEL'][:n+a+1]
            probas_train = model.predict_proba(pd.DataFrame(np.array(X_train_raw), columns=selected_features, index=X_train_raw.index))[:, 1]
            
            probas_test_list.append(probas_test)
            probas_train_list.append(probas_train)
            y_tests.append(y_test)
            y_trains.append(y_train)
            clf_names.append(model_name)
            
        except Exception as e:
            print(f"Error loading {model_name}: {e}")
    
    if len(probas_test_list) == 0:
        print("No successful calibrations!")
        return
    
    colors = sns.color_palette("tab10", len(probas_test_list))
    
    n_models = len(probas_test_list)
    # Set up subplot grid - 2 rows, 5 columns for 10 models
    nrows, ncols = 2, 5
    
    fig, axes = plt.subplots(nrows, ncols, figsize=(ncols*8, nrows*7))
    axes = axes.flatten()
    
    for i in range(n_models):
        # Calculate calibration curves
        prob_true_test, prob_pred_test = calibration_curve(y_tests[i], probas_test_list[i], n_bins=20)
        prob_true_train, prob_pred_train = calibration_curve(y_trains[i], probas_train_list[i], n_bins=20)
        
        ax = axes[i]
        
        # Plot test calibration curve
        ax.plot(prob_pred_test, prob_true_test, 'o-', color=colors[i], lw=2, label='Test', markersize=6)
        
        # Plot train calibration curve
        ax.plot(prob_pred_train, prob_true_train, 's--', color=colors[i], lw=2, alpha=0.7, label='Train', markersize=6)
        
        # Perfect calibration line
        ax.plot([0, 1], [0, 1], '--', color='gray', alpha=0.8)
        
        ax.set_title(f'{clf_names[i]}', fontsize=40, fontweight='bold')
        ax.set_xlabel('Mean Predicted Probability', fontsize=30, fontweight='bold')
        ax.set_ylabel('Fraction of Positives', fontsize=30, fontweight='bold')
        ax.tick_params(axis='both', which='major', labelsize=25)
        ax.legend(fontsize=30)
        ax.grid(True, alpha=0.5)
    
    # Hide unused subplots if any
    for i in range(n_models, nrows * ncols):
        axes[i].set_visible(False)
    
    # Add main title for the entire figure
    fig.suptitle('CSMSL+BFS+SMBO-GP Fusion', 
                 fontsize=50, fontweight='bold', y=1.02)
    
    # Add space between subplots
    plt.subplots_adjust(
        wspace=1.9,  # horizontal spacing between plots
        hspace=1.9,  # vertical spacing between plots
        bottom=0.05,   # space at bottom for legend
        top=1     # space at top for main title
    )
    plt.tight_layout()
    plt.savefig(f"/kaggle/working/calibration_curves_all_models_slide_{slide_num}.tiff", dpi=500, bbox_inches='tight')
    plt.show()
    
    # Print summary statistics
    print(f"\n📊 Calibration Summary for Slide {slide_num}:")
    print("="*60)
    
    for i, model_name in enumerate(clf_names):
        # Calculate Brier Score (lower is better)
        from sklearn.metrics import brier_score_loss
        test_brier = brier_score_loss(y_tests[i], probas_test_list[i])
        train_brier = brier_score_loss(y_trains[i], probas_train_list[i])
        
        print(f"{model_name:8s} - Test Brier: {test_brier:.4f}, Train Brier: {train_brier:.4f}")
    
    return probas_test_list, probas_train_list, y_tests, y_trains, clf_names

# Usage:
# Choose which slide to analyze (1-based indexing)
slide_to_analyze = slide_num  # Change this to analyze different slides (1 to num_slides)

probas_test_list, probas_train_list, y_tests, y_trains, clf_names = generate_calibration_plots_for_models(df, slide_to_analyze, n, nn)
```

```python
# 📈 All KS Statistic plots for Multiple Models on Specific Slide

import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import joblib
from scipy import stats
import pandas as pd
from sklearn.preprocessing import StandardScaler
import warnings
warnings.filterwarnings('ignore')

def generate_ks_plots_for_models(df, slide_num, n, nn):
    """
    Generate KS statistic plots for 10 different models on a specific slide
    
    Parameters:
    df: DataFrame with features and LABEL
    slide_num: Which slide to analyze (1-based indexing)
    n: Training data size parameter
    nn: Test data size parameter
    """
    print(f"Generating KS Statistic Plots for Slide {slide_num}...")
    
    # Model names and their corresponding file prefixes
    model_names = ['ET', 'RP', 'RotF', 'RRotF', 'AdaB', 'LB', 'GBM', 'XGB', 'CAT', 'LGB']
    
    probas_test_list = []
    probas_train_list = []
    y_tests = []
    y_trains = []
    clf_names = []
    
    # Initialize scaler
    scaler = StandardScaler()
    
    # Calculate slide indices
    a = nn * (slide_num - 1)  # Convert to 0-based indexing for calculation
    
    for model_name in model_names:
        try:
            # Load model and selected features
            model = joblib.load(f"/kaggle/working/model_{model_name}_slide_{slide_num}.pkl")
            selected_features = joblib.load(f"/kaggle/working/selected_features_{model_name}_slide_{slide_num}.pkl")
            
            # Test data
            X_test_raw = df[selected_features][n+a+1:n+a+1+nn]
            y_test = df['LABEL'][n+a+1:n+a+1+nn]
            X_test_scaled = pd.DataFrame(np.array(X_test_raw), 
                                       columns=selected_features, 
                                       index=X_test_raw.index)
            probas_test = model.predict_proba(X_test_scaled)
            
            # Handle both 1D and 2D probability arrays
            if probas_test.ndim == 2:
                probas_test = probas_test[:, 1]  # Get positive class probabilities
            # If 1D, assume it's already the positive class probabilities
            
            # Train data
            X_train_raw = df[selected_features][:n+a+1]
            y_train = df['LABEL'][:n+a+1]
            X_train_scaled = pd.DataFrame(np.array(X_train_raw), 
                                        columns=selected_features, 
                                        index=X_train_raw.index)
            probas_train = model.predict_proba(X_train_scaled)
            
            # Handle both 1D and 2D probability arrays
            if probas_train.ndim == 2:
                probas_train = probas_train[:, 1]  # Get positive class probabilities
            # If 1D, assume it's already the positive class probabilities
            
            probas_test_list.append(probas_test)
            probas_train_list.append(probas_train)
            y_tests.append(y_test)
            y_trains.append(y_train)
            clf_names.append(model_name)
            
        except Exception as e:
            print(f"Error loading {model_name}: {e}")
    
    if len(probas_test_list) == 0:
        print("No successful models!")
        return None, None, None, None, None
    
    n_models = len(probas_test_list)
    # Set up subplot grid - 2 rows, 5 columns for 10 models
    nrows, ncols = 2, 5
    
    fig, axes = plt.subplots(nrows, ncols, figsize=(ncols*9, nrows*7))
    axes = axes.flatten()
    
    # Keep track of the last used axis for legend placement
    last_ax = None
    
    for i in range(n_models):
        try:
            # Get probabilities for positive class (Buy) - Test data
            pos_probs_test = probas_test_list[i]
            sell_probs_test = pos_probs_test[y_tests[i] == 0]  # Class 0 = Sell
            buy_probs_test = pos_probs_test[y_tests[i] == 1]   # Class 1 = Buy
            
            # Get probabilities for positive class (Buy) - Train data
            pos_probs_train = probas_train_list[i]
            sell_probs_train = pos_probs_train[y_trains[i] == 0]  # Class 0 = Sell
            buy_probs_train = pos_probs_train[y_trains[i] == 1]   # Class 1 = Buy
            
            # Calculate KS statistics
            ks_stat_test, _ = stats.ks_2samp(buy_probs_test, sell_probs_test)
            ks_stat_train, _ = stats.ks_2samp(buy_probs_train, sell_probs_train)
            
            # Create threshold range
            thresholds = np.linspace(0, 1, 100)
            
            # Calculate cumulative distributions for TEST
            sell_cdf_test = []
            buy_cdf_test = []
            for threshold in thresholds:
                sell_cdf_test.append(np.mean(sell_probs_test <= threshold))
                buy_cdf_test.append(np.mean(buy_probs_test <= threshold))
            
            # Calculate cumulative distributions for TRAIN
            sell_cdf_train = []
            buy_cdf_train = []
            for threshold in thresholds:
                sell_cdf_train.append(np.mean(sell_probs_train <= threshold))
                buy_cdf_train.append(np.mean(buy_probs_train <= threshold))
            
            # Find maximum KS points
            ks_values_test = np.abs(np.array(buy_cdf_test) - np.array(sell_cdf_test))
            max_ks_idx_test = np.argmax(ks_values_test)
            max_ks_threshold_test = thresholds[max_ks_idx_test]
            
            ax = axes[i]
            last_ax = ax  # Keep track of the last axis used
            
            # Plot TEST cumulative distributions
            ax.plot(thresholds, sell_cdf_test, color='red', linewidth=2, label='Sell (Test)', alpha=0.8)
            ax.plot(thresholds, buy_cdf_test, color='green', linewidth=2, label='Buy (Test)', alpha=0.8)
            
            # Plot TRAIN cumulative distributions (dashed)
            ax.plot(thresholds, sell_cdf_train, color='red', linewidth=2, linestyle='--', label='Sell (Train)', alpha=0.6)
            ax.plot(thresholds, buy_cdf_train, color='green', linewidth=2, linestyle='--', label='Buy (Train)', alpha=0.6)
            
            # Add vertical line at maximum KS for test
            ax.axvline(x=max_ks_threshold_test, color='black', linestyle=':', linewidth=2, alpha=0.7)
            
            # Add KS statistics text
            ax.text(0.02, 0.98, f'KS Test: {ks_stat_test:.3f}\nKS Train: {ks_stat_train:.3f}', 
                   transform=ax.transAxes, fontsize=30,
                   bbox=dict(boxstyle="round,pad=0.3", facecolor="white", alpha=0.8),
                   verticalalignment='top')
            
            # Formatting
            ax.set_xlabel('Threshold', fontsize=35, fontweight='bold')
            ax.set_ylabel('Cumulative Percentage', fontsize=35, fontweight='bold')
            ax.set_title(f'{clf_names[i]}', fontsize=40, fontweight='bold')
            ax.grid(True, alpha=0.5)
            ax.set_xlim(0, 1)
            ax.set_ylim(0, 1)
            ax.tick_params(axis='both', which='major', labelsize=28)
            
            # Add percentage labels on y-axis
            ax.set_yticklabels([f'{int(y*100)}%' for y in ax.get_yticks()])
            
        except Exception as model_error:
            print(f"Error plotting {clf_names[i]}: {model_error}")
    
    # Hide unused subplots if any
    for i in range(n_models, nrows * ncols):
        axes[i].set_visible(False)
    
    # Add main title for the entire figure
    fig.suptitle('CSMSL+BFS Fusion', 
                 fontsize=50, fontweight='bold', y=1.05)
    
    # Add space between subplots
    plt.subplots_adjust(
        wspace=0.5,  # horizontal spacing between plots
        hspace=0.4,  # vertical spacing between plots
        bottom=0.001,   # space at bottom for legend
        top=0.93     # space at top for main title
    )
    
    # Add legend only if we have at least one successful plot
    if last_ax is not None:
        last_ax.legend(fontsize=40, loc='lower center', bbox_to_anchor=(-2.5, -0.5), 
                      ncol=4, frameon=False)
    
    plt.tight_layout()
    plt.savefig(f"/kaggle/working/ks_statistic_plots_all_models_slide_{slide_num}.tiff", dpi=500, bbox_inches='tight')
    plt.show()
    
    # Print summary statistics
    print(f"\n📊 KS Statistics Summary for Slide {slide_num}:")
    print("="*60)
    
    for i in range(len(probas_test_list)):
        try:
            # Calculate KS statistics for summary
            pos_probs_test = probas_test_list[i]
            sell_probs_test = pos_probs_test[y_tests[i] == 0]
            buy_probs_test = pos_probs_test[y_tests[i] == 1]
            
            pos_probs_train = probas_train_list[i]
            sell_probs_train = pos_probs_train[y_trains[i] == 0]
            buy_probs_train = pos_probs_train[y_trains[i] == 1]
            
            ks_stat_test, _ = stats.ks_2samp(buy_probs_test, sell_probs_test)
            ks_stat_train, _ = stats.ks_2samp(buy_probs_train, sell_probs_train)
            
            print(f"{clf_names[i]:8s} - KS Test: {ks_stat_test:.4f}, KS Train: {ks_stat_train:.4f}")
        except:
            continue

    return probas_test_list, probas_train_list, y_tests, y_trains, clf_names

# Usage:
# Choose which slide to analyze (1-based indexing)
slide_to_analyze = slide_num  # Change this to analyze different slides (1 to num_slides)

probas_test_list, probas_train_list, y_tests, y_trains, clf_names = generate_ks_plots_for_models(df, slide_to_analyze, n, nn)
```

```python
# 📈 Multi-Model Confusion Matrix for Single Slide with Multiple Metrics

import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np, scipy
import joblib
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
scipy.interp = np.interp 
import scikitplot as skplt
import warnings
warnings.filterwarnings('ignore')

def generate_confusion_matrix_for_models(df, slide_num, n, nn):
    """
    Generate confusion matrices for all 10 models for a specific slide
    
    Parameters:
    - df: DataFrame with features and LABEL
    - slide_num: Which slide to analyze (1-based indexing)
    - n: Training data size up to slide
    - nn: Test data size for slide
    """
    print(f"Generating Confusion Matrix Plots for Slide {slide_num}...")
    
    # Model names and their corresponding file prefixes
    model_names = ['ET', 'RP', 'RotF', 'RRotF', 'AdaB', 'LB', 'GBM', 'XGB', 'CAT', 'LGB']
    
    y_pred_list = []
    y_tests = []
    clf_names = []
    
    # Initialize scaler
    scaler = StandardScaler()
    
    # Calculate slide indices
    a = nn * (slide_num - 1)  # Convert to 0-based indexing for calculation
    
    for model_name in model_names:
        try:
            # Load model and selected features
            model = joblib.load(f"/kaggle/working/model_{model_name}_slide_{slide_num}.pkl")
            selected_features = joblib.load(f"/kaggle/working/selected_features_{model_name}_slide_{slide_num}.pkl")
            
            # Test data
            X_test_raw = df[selected_features][n+a+1:n+a+1+nn]
            y_test = df['LABEL'][n+a+1:n+a+1+nn]
            X_test_scaled = pd.DataFrame(np.array(X_test_raw), 
                                       columns=selected_features, 
                                       index=X_test_raw.index)
            
            # Use model.predict() which returns class predictions (1D array)
            y_pred = model.predict(X_test_scaled)
            
            # model.predict() typically returns a 1D array of predictions
            # No need for [:, 1] indexing since we want the actual predictions
            
            y_pred_list.append(y_pred)
            y_tests.append(y_test)
            clf_names.append(f"{model_name}")
            
            print(f"✓ Successfully processed {model_name}")
            
        except Exception as e:
            print(f"✗ Error processing {model_name}: {e}")
    
    if len(y_pred_list) == 0:
        print("No successful models!")
        return None, None, None
    
    n_models = len(y_pred_list)
    
    # Set up subplot layout for 10 models
    if n_models <= 6:
        nrows, ncols = 2, 3
    elif n_models <= 10:
        nrows, ncols = 2, 5
    else:
        nrows, ncols = 3, 4
    
    # Create figure with subplots
    fig, axes = plt.subplots(nrows, ncols, figsize=(ncols*8, nrows*8))
    if n_models == 1:
        axes = [axes]
    else:
        axes = axes.flatten()
    
    # Define model colors for consistency
    model_colors = {
        'ET': 'blue', 'RP': 'orange', 'RotF': 'green', 'RRotF': 'red',
        'AdaB': 'purple', 'LB': 'brown', 'GBM': 'pink', 'XGB': 'gray',
        'CAT': 'olive', 'LGB': 'cyan'
    }
    
    for i in range(n_models):
        try:
            ax = axes[i]
            model_name = clf_names[i]
            color = model_colors.get(model_name, 'black')
            
            # Plot confusion matrix with normalization
            skplt.metrics.plot_confusion_matrix(
                y_tests[i], 
                y_pred_list[i],
                normalize=True,
                ax=ax
            )
            # Remove the internal bars
            for c in ax.collections:
               c.remove()
            # Customize plot
            ax.set_title(f'{model_name}', fontsize=35, fontweight='bold')
            ax.set_xlabel('Predicted Label', fontsize=30, fontweight='bold')
            ax.set_ylabel('True Label', fontsize=30, fontweight='bold')
            ax.tick_params(axis='both', labelsize=30)
            
            # Customize the text annotations in the confusion matrix
            for text in ax.texts:
                text.set_fontsize(25)
                text.set_fontweight('bold')
            
            # Set class labels to be more descriptive
            ax.set_xticklabels(['Sell (0)', 'Buy (1)'], fontsize=22)
            ax.set_yticklabels(['Sell (0)', 'Buy (1)'], fontsize=22)
            
            # Calculate metrics
            accuracy = accuracy_score(y_tests[i], y_pred_list[i])
            precision = precision_score(y_tests[i], y_pred_list[i], average='weighted', zero_division=0)
            recall = recall_score(y_tests[i], y_pred_list[i], average='weighted', zero_division=0)
            f1 = f1_score(y_tests[i], y_pred_list[i], average='weighted', zero_division=0)
            
            # Add metrics in the four corners of the plot
            # Top-left corner: Accuracy
            ax.text(0.02, 0.98, f'Acc: {accuracy:.2f}', 
                   transform=ax.transAxes,
                   fontsize=20, fontweight='bold',
                   bbox=dict(boxstyle="round,pad=0.3", facecolor="lightblue", alpha=0.8),
                   verticalalignment='top', horizontalalignment='left')
            
            # Top-right corner: Precision
            ax.text(0.98, 0.98, f'Prec: {precision:.2f}', 
                   transform=ax.transAxes,
                   fontsize=20, fontweight='bold',
                   bbox=dict(boxstyle="round,pad=0.3", facecolor="lightgreen", alpha=0.8),
                   verticalalignment='top', horizontalalignment='right')
            
            # Bottom-left corner: Recall
            ax.text(0.02, 0.02, f'Rec: {recall:.2f}', 
                   transform=ax.transAxes,
                   fontsize=20, fontweight='bold',
                   bbox=dict(boxstyle="round,pad=0.3", facecolor="lightyellow", alpha=0.8),
                   verticalalignment='bottom', horizontalalignment='left')
            
            # Bottom-right corner: F1 Score
            ax.text(0.98, 0.02, f'F1: {f1:.2f}', 
                   transform=ax.transAxes,
                   fontsize=20, fontweight='bold',
                   bbox=dict(boxstyle="round,pad=0.3", facecolor="lightcoral", alpha=0.8),
                   verticalalignment='bottom', horizontalalignment='right')
                
        except Exception as model_error:
            print(f"Error plotting {clf_names[i]}: {model_error}")
            axes[i].text(0.5, 0.5, f'Error: {clf_names[i]}', 
                        ha='center', va='center', transform=axes[i].transAxes,
                        fontsize=25)
    
    # Hide unused subplots
    for i in range(n_models, len(axes)):
        axes[i].set_visible(False)
    
    # Add space between subplots
    plt.subplots_adjust(
        wspace=0.5,  # horizontal spacing between plots
        hspace=0.4,  # vertical spacing between plots
        top=0.65,    # space at top for main title
        bottom=0.05  # space at bottom
    )
    
    # Add overall title
    # Add main title for the entire figure
    fig.suptitle('CSMSL+BFS Fusion', 
                 fontsize=50, fontweight='bold', y=0.75)
    
    plt.savefig(f"/kaggle/working/confusion_matrix_slide_{slide_num}_all_models.tiff", dpi=500, bbox_inches='tight')
    plt.show()
    
    return y_pred_list, y_tests, clf_names

# Usage:
y_pred_list, y_tests, clf_names = generate_confusion_matrix_for_models(df, slide_num=slide_num, n=n, nn=nn)
```

```python
# 📈 Multi-Model Metrics Bar Plots for Single Slide

import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import joblib
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, precision_score, recall_score, matthews_corrcoef
import warnings
warnings.filterwarnings('ignore')

def generate_metrics_bar_plots_for_models(df, slide_num, n, nn):
    """
    Generate metrics bar plots for all 10 models for a specific slide
    
    Parameters:
    - df: DataFrame with features and LABEL
    - slide_num: Which slide to analyze (1-based indexing)
    - n: Training data size up to slide
    - nn: Test data size for slide
    """
    print(f"Generating Metrics Bar Plots for Slide {slide_num}...")
    
    # Model names and their corresponding file prefixes
    model_names = ['ET', 'RP', 'RotF', 'RRotF', 'AdaB', 'LB', 'GBM', 'XGB', 'CAT', 'LGB']
    
    models_metrics = []
    clf_names = []
    
    # Initialize scaler
    scaler = StandardScaler()
    
    # Calculate slide indices
    a = nn * (slide_num - 1)  # Convert to 0-based indexing for calculation
    
    for model_name in model_names:
        try:
            # Load model and selected features
            model = joblib.load(f"/kaggle/working/model_{model_name}_slide_{slide_num}.pkl")
            selected_features = joblib.load(f"/kaggle/working/selected_features_{model_name}_slide_{slide_num}.pkl")
            
            # Test data only
            X_test_raw = df[selected_features][n+a+1:n+a+1+nn]
            y_test = df['LABEL'][n+a+1:n+a+1+nn]
            X_test_scaled = pd.DataFrame(np.array(X_test_raw), 
                                       columns=selected_features, 
                                       index=X_test_raw.index)
            
            # Use model.predict() which returns class predictions (1D array)
            y_pred = model.predict(X_test_scaled)
            # No need for [:, 1] indexing since model.predict() returns actual predictions
            
            # Calculate metrics
            accuracy = accuracy_score(y_test, y_pred)
            precision = precision_score(y_test, y_pred, average='weighted', zero_division=0)
            recall = recall_score(y_test, y_pred, average='weighted', zero_division=0)
            mcc = matthews_corrcoef(y_test, y_pred)
            
            # Store metrics for this model
            models_metrics.append([accuracy, precision, recall, mcc])
            clf_names.append(f"{model_name}")
            
            print(f"✓ {model_name} - Acc: {accuracy:.3f}, Prec: {precision:.3f}, Rec: {recall:.3f}, MCC: {mcc:.3f}")
            
        except Exception as e:
            print(f"✗ Error processing {model_name}: {e}")
    
    if len(models_metrics) == 0:
        print("No successful models!")
        return None, None
    
    n_models = len(models_metrics)
    
    # Set up subplot layout for 10 models
    if n_models <= 6:
        nrows, ncols = 2, 3
    elif n_models <= 10:
        nrows, ncols = 2, 5
    else:
        nrows, ncols = 3, 4
    
    # Create figure with subplots
    fig, axes = plt.subplots(nrows, ncols, figsize=(ncols*7, nrows*6))
    if n_models == 1:
        axes = [axes]
    else:
        axes = axes.flatten()
    
    # Define model colors for consistency
    model_colors = {
        'ET': 'blue', 'RP': 'orange', 'RotF': 'green', 'RRotF': 'red',
        'AdaB': 'purple', 'LB': 'brown', 'GBM': 'pink', 'XGB': 'gray',
        'CAT': 'olive', 'LGB': 'cyan'
    }
    
    metric_names = ['Accuracy', 'Precision', 'Recall', 'MCC']
    colors = ['skyblue', 'lightgreen', 'lightcoral', 'lightsalmon']
    
    for i in range(n_models):
        try:
            ax = axes[i]
            model_name = clf_names[i]
            title_color = model_colors.get(model_name, 'black')
            
            # Get metrics for this model
            metrics = models_metrics[i]
            
            # Create bar positions
            x_pos = np.arange(len(metric_names))
            
            # Create bars
            bars = ax.bar(x_pos, metrics, color=colors, alpha=0.7, 
                         edgecolor='black', linewidth=1)
            
            # Add value labels on top of bars
            for bar, value in zip(bars, metrics):
                height = bar.get_height()
                # Handle negative values (MCC can be negative)
                label_y = height + 0.02 if height >= 0 else height - 0.05
                ax.text(bar.get_x() + bar.get_width()/2., label_y,
                       f'{value:.2f}', ha='center', 
                       va='bottom' if height >= 0 else 'top', 
                       fontsize=25, fontweight='bold')
            
            # Customize plot - REMOVED X-AXIS LABELS
            ax.set_title(f'{model_name}', fontsize=35, fontweight='bold')
            ax.set_ylabel('Score', fontsize=30, fontweight='bold')
            ax.set_xticks(x_pos)
            ax.set_xticklabels([])  # Empty x-axis labels
            ax.tick_params(axis='y', labelsize=25)
            ax.tick_params(axis='x', length=0)  # Remove x-axis tick marks
            ax.grid(True, alpha=0.3, axis='y')
            
            # Set y-limits to accommodate MCC range (-1 to 1) and other metrics (0 to 1)
            y_min = min(0, min(metrics) - 0.1)
            y_max = max(1, max(metrics) + 0.1)
            ax.set_ylim([y_min, y_max])
            
            # Add horizontal line at y=0 for reference
            ax.axhline(y=0, color='black', linestyle='-', alpha=0.3)
                
        except Exception as model_error:
            print(f"Error plotting {clf_names[i]}: {model_error}")
            axes[i].text(0.5, 0.5, f'Error: {clf_names[i]}', 
                        ha='center', va='center', transform=axes[i].transAxes,
                        fontsize=25)
    
    # Hide unused subplots
    for i in range(n_models, len(axes)):
        axes[i].set_visible(False)
    
    # Add space between subplots
    plt.subplots_adjust(
        wspace=0.3,  # horizontal spacing between plots
        hspace=0.3,  # vertical spacing between plots
        top=0.92,    # space at top for main title
        bottom=0.15   # space at bottom for legend
    )
    
    # Create global legend at bottom with 4 columns
    from matplotlib.patches import Patch
    legend_elements = [
        Patch(facecolor='skyblue', edgecolor='black', label='Accuracy'),
        Patch(facecolor='lightgreen', edgecolor='black', label='Precision'),
        Patch(facecolor='lightcoral', edgecolor='black', label='Recall'),
        Patch(facecolor='lightsalmon', edgecolor='black', label='MCC'),
    ]
    fig.legend(handles=legend_elements, fontsize=35, loc='lower center', 
               bbox_to_anchor=(0.5, 0.02), ncol=4, frameon=False)
    
    # Add main title for the entire figure
    fig.suptitle('CSMSL+BFS Fusion', 
                 fontsize=50, fontweight='bold', y=1.05)
    
    plt.savefig(f"/kaggle/working/metrics_bar_plots_slide_{slide_num}_all_models.tiff", dpi=500, bbox_inches='tight')
    plt.show()
    
    return models_metrics, clf_names

# Usage:
models_metrics, clf_names = generate_metrics_bar_plots_for_models(df, slide_num=slide_num, n=n, nn=nn)
```

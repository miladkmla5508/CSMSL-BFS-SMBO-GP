```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

# Define the file name
file_path = '/kaggle/input/datasets/miladkamali/fusion-ablation/CSMSLBFSSMBO-GP trading ablation.xlsx'

# Define the sheet names and their corresponding subplot titles
sheets = {
    'BTC':   '1-Min BTC/USDT',
    'ETH':   '1-Min ETH/USDT',
    'SOL':   '1-Min SOL/USDT',
    'XRP':   '1-Min XRP/USDT',
    'DOGE':  '1-Min DOGE/USDT',
    'LTC':   '1-Min LTC/USDT'
}

# Define the column names for plotting
columns = [
    'Without Fusion',
    'BFS Fusion',
    'CSMSL+BFS Fusion',
    'CSMSL+BFS+SMBO-GP Fusion'
]

# ---- CUSTOM COLORS ----
colors = {
    'Without Fusion':             '#e6194b',   # red
    'BFS Fusion':                 '#f58231',   # orange
    'CSMSL+BFS Fusion':           '#3cb44b',   # green
    'CSMSL+BFS+SMBO-GP Fusion':   '#4363d8',   # blue
}

# Create a figure and a 2x3 grid of subplots
fig, axes = plt.subplots(2, 3, figsize=(25, 15), sharex=False)
axes = axes.flatten()

# Loop through each sheet, read the data, and plot it
for i, (sheet_name, title) in enumerate(sheets.items()):
    df = pd.read_excel(file_path, sheet_name=sheet_name, parse_dates=['Date'])
    ax = axes[i]

    for col in columns:
        ax.plot(
            df['Date'], df[col],
            label=col,
            color=colors[col],
            linewidth=2
        )

    # ---- BIG TITLE ----
    ax.set_title(title, fontsize=30, weight='bold', pad=12)

    # ---- BIG AXIS LABELS ----
    ax.set_ylabel('Balance', fontsize=25, weight='bold')

    # ---- BIG TICK LABELS (pointing INWARD, longer, on all 4 sides) ----
    ax.tick_params(
        axis='both',
        which='major',
        direction='inout',      # ticks point INTO the plot
        length=12,           # longer ticks
        width=2,             # thicker ticks
        labelsize=20,        # big tick labels
        top=False,            # ticks on top edge too
        right=False,          # ticks on right edge too
    )
    plt.setp(ax.get_xticklabels(), rotation=45, ha='right')

    # ---- DATE + TIME FORMATTING (2-digit year) ----
    ax.xaxis.set_major_locator(mdates.AutoDateLocator())
    ax.xaxis.set_major_formatter(mdates.DateFormatter('%y-%m-%d %H:%M'))

    # ---- GRID ----
    ax.grid(True, which='both', linestyle='--', linewidth=0.6, alpha=0.7)

# --- Single shared legend at the bottom ---
handles, labels = axes[0].get_legend_handles_labels()
fig.legend(
    handles, labels,
    loc='lower center',
    bbox_to_anchor=(0.5, -0.01),
    ncol=4,
    fontsize=30,
    frameon=True,
    title='',
    title_fontsize=16
)

fig.tight_layout(rect=[0, 0.06, 1, 0.98])

plt.savefig('trading_ablation_subplots.svg', dpi=500, bbox_inches='tight')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

# Define the file name
file_path = '/kaggle/input/datasets/miladkamali/trading-ablation/Swing trading ablation.xlsx'

# Define the sheet names and their corresponding subplot titles
sheets = {
    'BTC':  '1-Min BTC/USDT',
    'ETH':  '1-Min ETH/USDT',
    'SOL':  '1-Min SOL/USDT',
    'XRP':  '1-Min XRP/USDT',
    'DOGE': '1-Min DOGE/USDT',   
    'LTC':  '1-Min LTC/USDT'
}

# Define the column names for plotting
columns = [
    'Short & Hold', 'Long & Hold', 'Momentum', 'Mean Reversion',
    'Breakout', 'Short Only', 'Long Only', 'Swing'
]

# ---- CUSTOM COLORS (8 distinct colors) ----
colors = {
    'Short & Hold':   '#e6194b',  # red
    'Long & Hold':    '#f58231',  # orange
    'Momentum':       '#d4b800',  # darker yellow
    'Mean Reversion': '#bfef45',  # lime
    'Breakout':       '#3cb44b',  # green
    'Short Only':     '#42d4f4',  # cyan
    'Long Only':      '#4363d8',  # blue
    'Swing':          '#911eb4',  # purple
}

# Create a figure and a 2x3 grid of subplots
fig, axes = plt.subplots(2, 3, figsize=(30, 20), sharex=False)
axes = axes.flatten()

# Loop through each sheet, read the data, and plot it
for i, (sheet_name, title) in enumerate(sheets.items()):
    df = pd.read_excel(file_path, sheet_name=sheet_name, parse_dates=['date'])

    # ---- strip trailing/leading whitespace from column names ----
    # fixes 'Momentum  ' and 'Breakout  ' in the workbook
    df.columns = df.columns.str.strip()

    ax = axes[i]

    for col in columns:
        ax.plot(
            df['date'], df[col],
            label=col,
            color=colors[col],
            linewidth=2
        )

    # ---- BIG TITLE ----
    ax.set_title(title, fontsize=35, weight='bold', pad=15)

    # ---- BIG AXIS LABELS ----
    ax.set_ylabel('Balance', fontsize=30, weight='bold')

    # ---- BIG TICK LABELS (pointing INWARD, longer, on all 4 sides) ----
    ax.tick_params(
        axis='both',
        which='major',
        direction='inout',      # ticks point INTO the plot
        length=12,           # longer ticks
        width=2,             # thicker ticks
        labelsize=25,        # big tick labels
        top=False,            # ticks on top edge too
        right=False,          # ticks on right edge too
    )
    plt.setp(ax.get_xticklabels(), rotation=45, ha='right')

    # ---- DATE + TIME FORMATTING (2-digit year) ----
    ax.xaxis.set_major_locator(mdates.AutoDateLocator())
    ax.xaxis.set_major_formatter(mdates.DateFormatter('%y-%m-%d %H:%M'))

    ax.grid(True, which='both', linestyle='--', linewidth=0.6, alpha=0.7)

# Hide unused subplots if fewer than 6 sheets are present
for j in range(len(sheets), len(axes)):
    axes[j].set_visible(False)

# --- Single shared legend at the bottom ---
handles, labels = axes[0].get_legend_handles_labels()
fig.legend(
    handles, labels,
    loc='lower center',
    bbox_to_anchor=(0.5, -0.03),
    ncol=4,             # 4 cols × 2 rows for 8 strategies
    fontsize=40,
    frameon=True,
    title='',
)

fig.tight_layout(rect=[0, 0.08, 1, 0.97])

plt.savefig('swing_trading_ablation_subplots.svg', dpi=500, bbox_inches='tight')
plt.show()
```
